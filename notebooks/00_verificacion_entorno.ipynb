# 00. Verificación del entorno
Este notebook comprueba el entorno, las carpetas y la escritura de archivos antes de descargar datos científicos.

```python
from pathlib import Path
import sys, platform, json
import numpy as np
import pandas as pd
import scipy
import sklearn

print('Python:', sys.version)
print('Ejecutable:', sys.executable)
print('Sistema:', platform.platform())
print('NumPy:', np.__version__)
print('pandas:', pd.__version__)
print('SciPy:', scipy.__version__)
print('scikit-learn:', sklearn.__version__)
```

```python
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
required = ['data/raw','data/interim','data/processed','data/metadata','results/figures','results/tables','results/diagnostics']
for rel in required:
    p = ROOT / rel
    p.mkdir(parents=True, exist_ok=True)
    print('OK', p)
```

```python
test = pd.DataFrame({'pais':['ECU','MEX','FIN'], 'valor':[1.0,2.0,3.0]})
out = ROOT / 'data/interim/prueba_notebook.parquet'
test.to_parquet(out, index=False)
check = pd.read_parquet(out)
assert test.equals(check)
print('PRUEBA COMPLETA:', out)
```
